```python
# Install spaCy
!pip install spacy

# Download English model
!python -m spacy download en_core_web_sm

# Import spaCy
import spacy

# Load model
nlp = spacy.load("en_core_web_sm")

print("spaCy model loaded successfully!")
```

```python
sentences = [
    "The cat sits on the mat.",
    "She is reading a book.",
    "He went to the market and bought vegetables.",
    "Although it was raining, we went outside.",
    "Do you like pizza?",
    "The project was completed by the team.",
    "Artificial intelligence is transforming industries.",
    "I posted a photo on Instagram yesterday.",
    "The server crashed due to high traffic.",
    "She sings beautifully.",
    "They are playing football in the ground.",
    "The teacher explained the concept clearly.",
    "While I was studying, my phone rang.",
    "The cake was baked by my mother.",
    "Can you solve this problem?"
]
```

```python
import pandas as pd

all_data = []

for sentence in sentences:
    doc = nlp(sentence)
    for token in doc:
        all_data.append([
            sentence,
            token.text,
            token.lemma_,
            token.pos_,
            token.dep_,
            token.head.text
        ])

# Create DataFrame
df = pd.DataFrame(all_data, columns=[
    "Sentence", "Token", "Lemma", "POS", "Dependency", "Head"
])

df.head(20)
```

```python
for sentence in sentences:
    doc = nlp(sentence)
    print("\nSentence:", sentence)

    subject = ""
    verb = ""
    obj = ""

    for token in doc:
        if "subj" in token.dep_:
            subject = token.text
        if token.dep_ == "ROOT":
            verb = token.text
        if "obj" in token.dep_:
            obj = token.text

    print("Subject:", subject)
    print("Verb (ROOT):", verb)
    print("Object:", obj)
```

```python
from spacy import displacy

# Visualize first 5 sentences
for sentence in sentences[:5]:
    doc = nlp(sentence)
    displacy.render(doc, style="dep", jupyter=True)
```

```python
analysis = """
1. Simple sentences like 'The cat sits on the mat' have straightforward structures.
2. Compound sentences show multiple clauses connected by conjunctions.
3. Complex sentences include dependent clauses, making trees deeper.
4. Interrogative sentences often invert subject and auxiliary verbs.
5. Passive voice changes subject-object roles.
6. Modifiers increase tree complexity.
7. Technical sentences show structured dependencies.
8. Social sentences are simpler and direct.
"""

print(analysis)
```
